```python
import cvxpy as cp
import numpy as np

# ==============================================================================
# Chemical Equilibrium Problem (White, Johnson, Dantzig 1958)
# Mixture: 1/2 N2H4 + 1/2 O2 at 3500K, 750 psi
# ==============================================================================

n_compounds = 10
n_elements = 3

compounds = ["H", "H2", "H2O", "N", "N2", "NH", "NO", "O", "O2", "OH"]

# c_j = (F0/RT)_j + ln(P), where P = 750 psi
c = np.array([
    -6.089,   # H
    -17.164,  # H2
    -34.054,  # H2O
    -5.914,   # N
    -24.721,  # N2
    -14.986,  # NH
    -24.100,  # NO
    -10.708,  # O
    -26.662,  # O2
    -22.179   # OH
])

# Stoichiometry matrix A[i,j] = atoms of element i in compound j
# Rows: H, N, O
A = np.array([
    [1, 2, 2, 0, 0, 1, 0, 0, 0, 1],  # H
    [0, 0, 0, 1, 2, 1, 1, 0, 0, 0],  # N
    [0, 0, 1, 0, 0, 0, 1, 1, 2, 1],  # O (H2O has 1 oxygen atom)
], dtype=float)

# Atomic weights in mixture
b = np.array([2.0, 1.0, 1.0])

# ==============================================================================
# CVXPY Formulation
# ==============================================================================

# Decision variables: moles of each compound
x = cp.Variable(n_compounds, nonneg=True)

# Total moles
x_total = cp.sum(x)

# Objective: minimize Gibbs free energy
# f(x) = sum_j x_j * (c_j + ln(x_j / x_total))
#      = c @ x + sum_j x_j * ln(x_j) - x_total * ln(x_total)
#      = c @ x - sum(entr(x)) + entr(x_total)
objective = c @ x - cp.sum(cp.entr(x)) + cp.entr(x_total)

# Mass balance constraints
constraints = [A @ x == b]

# Problem
prob = cp.Problem(cp.Minimize(objective), constraints)
prob.solve(solver=cp.IPOPT, nlp=True, verbose=True)

# Output results
print("Optimal value:", prob.value)
for i in range(n_compounds):
    print(f"{compounds[i]}: {x.value[i]:.4f} moles")
    
```

```python
import plotly.graph_objects as go
import plotly.express as px

# Extract solution
moles = x.value
total_moles = np.sum(moles)

# Create nicely formatted labels with subscripts
compound_labels = ["H", "H<sub>2</sub>", "H<sub>2</sub>O", "N", "N<sub>2</sub>", 
                   "NH", "NO", "O", "O<sub>2</sub>", "OH"]

# Sort by moles (decreasing order)
sorted_indices = np.argsort(moles)[::-1]
sorted_labels = [compound_labels[i] for i in sorted_indices]
sorted_moles = moles[sorted_indices]

# Generate colors using a different color scheme
# Try different color palettes by uncommenting one of these:
# colors = px.colors.qualitative.Plotly[:n_compounds]       # Bright and distinct
# colors = px.colors.qualitative.D3[:n_compounds]           # D3.js default colors
# colors = px.colors.qualitative.G10[:n_compounds]          # Tableau 10 colors
# colors = px.colors.qualitative.T10[:n_compounds]          # Another Tableau variant
# colors = px.colors.qualitative.Alphabet[:n_compounds]     # Large color set
# colors = px.colors.qualitative.Dark24[:n_compounds]       # Darker tones
# colors = px.colors.qualitative.Light24[:n_compounds]      # Lighter tones
# colors = px.colors.qualitative.Set1[:n_compounds]         # ColorBrewer Set1
# colors = px.colors.qualitative.Set2[:n_compounds]         # ColorBrewer Set2
# colors = px.colors.qualitative.Set3[:n_compounds]         # ColorBrewer Set3
colors = px.colors.qualitative.Pastel[:n_compounds]        # Soft pastel colors
# colors = px.colors.qualitative.Bold[:n_compounds]         # Bold colors
# colors = px.colors.qualitative.Vivid[:n_compounds]        # Vivid colors
# colors = px.colors.qualitative.Safe[:n_compounds]         # Colorblind-safe

# Create bar chart with logarithmic scale
fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=sorted_labels,
        y=sorted_moles,
        marker=dict(
            color=colors
        ),
        hovertemplate='<b>%{x}</b><br>Moles: %{y:.4e}<extra></extra>'
    )
)

# Update layout with log scale and improved styling
fig.update_layout(
    xaxis={
        'title': 'Chemical Species',
        'title_font': {'size': 24},
        'tickfont': {'size': 22}
    },
    yaxis={
        'title': 'Moles',
        'title_font': {'size': 24},
        'type': 'log',
        'tickfont': {'size': 22},
        'gridcolor': 'rgba(128,128,128,0.2)',
        'tickvals': [1e-3, 1e-2, 1e-1, 1],
        'ticktext': ['1e-3', '1e-2', '1e-1', '1']
    },
    height=600,
    width=1000,
    plot_bgcolor='white',
    paper_bgcolor='white',
    hovermode='x unified'
)

# Save to PDF file
output_file = 'chemical_equilibrium_plot.pdf'
fig.write_image(output_file)
print(f"Plot saved to {output_file}")

# Print summary
print(f"\nTotal moles at equilibrium: {total_moles:.4f}")
print(f"Gibbs free energy (minimized): {prob.value:.4f}")

# Top 3 species by moles
top3_indices = np.argsort(moles)[-3:][::-1]
print(f"\nTop 3 species by moles:")
for idx in top3_indices:
    print(f"  {compounds[idx]:5s}: {moles[idx]:.4f} moles")

print(f"\nMoles of each species:")
for compound, mol in zip(compounds, moles):
    print(f"  {compound:5s}: {mol:.4e} moles")

# Display figure
fig
```
